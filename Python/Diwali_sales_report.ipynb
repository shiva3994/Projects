### Missing value and Anamolous detection

```python
df.isnull()
# check for null
```

```python
df.isnull().sum()
# check for sum of null columnwise
```

```python
null_counts = df.isnull().sum()
null_counts
```

```python
null_counts = null_counts[null_counts > 0]
null_counts
# show column containing nullhaving value above 0
```

```python
plt.figure(figsize=(10,5))
null_counts.plot(kind='bar',color='green')

plt.title('Missing value per column') # title
plt.xlabel('Column name') # X axis name
plt.ylabel('Number of missing values') # Y axis name

plt.xticks(rotation = 40)
plt.yticks(rotation = 40)

plt.show() # show final graph 
```

```python
cat_cols = ['Status','unnamed1']
for col in cat_cols:
    df[col] = df[col].fillna('unknown')
# Resolve the categorical with unknown

num_cols = ['Amount']
for col in num_cols:
    df[col] = df[col].fillna(df[col].median())
# Resolve the numerical with median
```

```python
# Manual Effort
# Plotting a bar chart for Gender and it's count

plt.figure(figsize=(8,5))

counts = df['Gender'].value_counts()

plt.bar(counts.index,counts.values, color = 'seagreen')

plt.xlabel('Gender')
plt.ylabel('Count')
plt.title('Gender Distribution')

plt.show()
```

```python
# Automation
# Plotting a bar chart for Gender and it's count

counts_a = sns.countplot(x = 'Gender', data = df, color = 'orange')
# sns.countplot(): Creates a bar chart using Seaborn that counts occurrences of each unique value in the specified column.
# x = 'Gender': Places the distinct values from the 'Gender' column (e.g., "Male", "Female") on the horizontal X-axis.
# data = df: Tells Seaborn to pull the 'Gender' column from your pandas DataFrame named df.
# counts_a = .: Stores the created plot object (a matplotlib Axes) into the variable counts_a.


for bars in counts_a.containers:
    counts_a.bar_label(bars)
# "Go through every group of bars on my chart (counts_a.containers), and slap a clear number label on top of each individual bar (counts_a.bar_label)."
```

```python
# Plotting a bar chart for gender vs total amount

sales_by_gender = df.groupby(['Gender'],as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False)


sns.barplot(x = 'Gender', y = 'Amount', data = sales_by_gender, color = 'Orange')
```

```python
sales_by_age = sns.countplot(data = df, x = 'Age Group', hue = 'Gender')

for bars in sales_by_age.containers:
    sales_by_age.bar_label(bars)
```
