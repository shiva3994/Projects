##### **Type B — GroupBy + filter (pandas' HAVING equivalent)**

##### 6. Find departments where average `Salary` exceeds 60000 (`.groupby().filter()` or `.groupby().mean()` + boolean mask).

```python
avg_salary_by_dept = data_dict['Employees'].groupby('Department', as_index=False)['Salary'].mean()

high_salary_depts = avg_salary_by_dept[avg_salary_by_dept['Salary'] > 60000]

high_salary_depts
```

##### 7. Find industries where total `ContractValue` exceeds 100000000.


```python
print(list(data_dict['Clients'].columns))
```

```python
total_ctcvalue_by_in = data_dict['Clients'].groupby('Industry', as_index=False)['ContractValue'].sum()

high_ctcvalue = total_ctcvalue_by_in[total_ctcvalue_by_in['ContractValue'] > 100000000]

high_ctcvalue
```

##### 8. Find project statuses where average `Budget` exceeds 5000000.


```python
print(list(data_dict['Projects'].columns))
```

```python
avg_budget = data_dict['Projects'].groupby('Status',as_index=False)['Budget'].mean()

high_budget = avg_budget[avg_budget['Budget'] > 5000000]

high_budget
```

##### 9. Find companies where total `Revenue` exceeds 50000.


```python
print(list(data_dict['Company_Financials'].columns))
```

```python
total_rev = data_dict['Company_Financials'].groupby('Company',as_index=False)['Revenue'].sum()

high_rev = total_rev[total_rev['Revenue'] > 50000]

high_rev
```
