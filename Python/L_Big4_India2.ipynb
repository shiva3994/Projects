##### **Type A — `.rank()` within groups**


##### 1. Rank employees within each `Department` by `Salary` descending using `.groupby()['Salary'].rank()`.


```python
print(list(data_dict['Employees'].columns))
```

```python
data_dict['Employees']['Salary_Rank'] = (
    data_dict['Employees']
    .groupby('Department')['Salary']
    .rank(ascending=False, method='dense')
)

data_dict['Employees'].head()
```

##### **Type C — `.cumsum()` running totals**


##### 11. Calculate a running total of `Salary` ordered by `Employee_ID`.


```python
data_dict['Employees'] = data_dict['Employees'].sort_values('Employee_ID')
data_dict['Employees']['running_total'] = data_dict['Employees']['Salary'].cumsum()

data_dict['Employees'].head()
```

##### 12. Calculate a running total of `ContractValue` ordered by `Client_ID`.


```python
data_dict['Clients'] = data_dict['Clients'].sort_values('Client_ID')
data_dict['Clients']['running_Clients'] = data_dict['Clients']['ContractValue'].cumsum()

data_dict['Clients'].head()
```

##### 13. Calculate a running total of `Budget` ordered by `Project_ID`.

```python
print(list(data_dict['Projects'].columns))
```
