##### **Simple Calculations**

##### 11. Find the average CO2 emission value for the year `2019` across all countries.

```python
df.head()
```

```python
avg_emission = df['2019'].mean()
print(f'Average CO2 emission for 2019 is : {avg_emission:.2f}')
# f prefix tells Python this is an f-string, which lets you embed variable values directly inside a string using curly braces {}
# .2f means "format as a fixed-point (decimal) number, rounded to 2 decimal places.

# {value:.0f} → rounds to a whole number, no decimals
# {value:,.2f} → adds comma separators for thousands (e.g. 1,234.56)
# {value:.2%} → formats a decimal as a percentage (e.g. 0.418 becomes 41.80%)
```

##### 12. Find which country had the highest CO2 emission in `2019`.

```python
df.head()
```

```python
highest_country_2019 = df.loc[df['2019'].idxmax(), 'Country Name']
print(f'Country with highest CO2 emission in 2019 is {highest_country_2019}')

# df['2019'].idxmax() — finds the index label of the row with the maximum value in the 2019 column (not the value itself, but where it is).
# df.loc[..., 'Country'] — uses that index to look up the corresponding value in the Country column.
```

##### 13. Find which country had the lowest CO2 emission in `2019` (excluding missing values).

```python
df.head()
```

```python
Lowest_CO2_2019 = df.loc[df['2019'].idxmin(), 'Country Name']
print(f'Lowest CO2 emission for a country was {Lowest_CO2_2019}')

#idxmin() (like idxmax(), mean(), min(), max(), etc.) automatically excludes NaN values by default in pandas.


# One edge case to be aware of: if the 2019 column is entirely NaN (no valid values at all),
# idxmin() will raise a ValueError since there's nothing to compare.
# If you want to be extra explicit about excluding missing values (e.g., for a teacher or portfolio reviewer to see the intent clearly),
# you could optionally write:

# valid_2019 = df.dropna(subset=['2019'])
# Lowest_CO2_2019 = valid_2019.loc[valid_2019['2019'].idxmin(), 'Country Name']
# print(f'Lowest CO2 emission for a country was {Lowest_CO2_2019}')
```

##### **Simple Visualization**

##### 20. Plot a bar chart of the top 10 countries by 2019 CO2 emissions using matplotlib.

```python
top10_2019 = df.sort_values(by='2019', ascending=False).head(10)

plt.figure(figsize=(18,6))

plt.bar(top10_2019['country_name'], top10_2019['2019'], color='orange')
plt.xlabel('Country')
plt.ylabel('CO2 Emissions (2019)')
plt.title('Top 10 Countries by 2019 CO2 Emissions')
plt.xticks(rotation=45)
plt.show()
```
