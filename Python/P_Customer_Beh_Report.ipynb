```python
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))

# .transform() keeps the original size of your DataFrame and do not shrinks the same.
# The lambda keyword creates a temporary, one-line function
# x represents the data for a single group (for example, all the rows belonging to the "Electronics" category).
# x.median() calculates the middle value (median) for that specific group's data.
# It looks for empty cells (NaN) only within that current group and fills them with the median calculated.
```

```python
df.isnull().sum()
```

```python
df.columns = df.columns.str.lower()
# convert columns name to lower

df.columns = df.columns.str.replace(' ','_')
# remove any spaces and replace it with _*

df = df.rename(columns = {'purchase_amount_(usd)':'purchase_amount'})
```

##### Creat a new column age_group

```python
labels = ['Young Adult', 'Adult', 'Middle-aged', 'Senior']

df['age_group'] = pd.qcut(df['age'], q = 4, labels = labels)
                # qcut splits age in 4 equal size data groups and assign the labels assigned
```

##### Create new column purchase_frequency_days  

```python
frequency_mapping = {
    'Fortnightly' : 14,
    'Weekly' : 7,
    'Mopnthly' : 30,
    'Quarterly' : 90,
    'Bi-Weekly' : 14,
    'Annually' : 365,
    'Every 3 Months' : 90
}


df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```
