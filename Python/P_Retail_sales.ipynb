##### Check the business rules: does Profit equal Sales_Amount minus Cost_Amount on every row? Do Return_Flag and Order_Status agree? List the mismatches and decide how to treat them.  

##### ========================== ACTION 1 ==========================

```python
# Calculate what Profit SHOULD be by subtracting Cost_Amount from Sales_Amount.

# WHY:
# In accounting and business logic: Profit = Revenue (Sales) - Cost.
# We round to 2 decimal places to match currency format and avoid false alerts 
# from tiny Python floating-point precision quirks (e.g., 0.1 + 0.2 = 0.30000000000000004).
# ------------------------------------------------------------------------------
# ------------------------------------------------------------------------------
expected_profit = (df_transactions['Sales_Amount'] - df_transactions['Cost_Amount']).round(2)

expected_profit.head(5)
```

```python
# Find the difference between the recorded 'Profit' in the dataset and our expected profit.

# WHY:
# If difference == 0, the row is accurate.
# If difference != 0, there is a discrepancy we need to investigate.
# ------------------------------------------------------------------------------
profit_diff = (df_transactions['Profit'] - expected_profit).round(2)

profit_diff.head(5)
```

```python
# Categorize the discrepancies into:
#   1. Exact matches (diff == 0)
#   2. Penny rounding differences (abs(diff) == 0.01)
#   3. Severe errors (abs(diff) > 0.01)

# WHY:
# - A 1-cent difference is very common in financial systems when rounding tax/discounts
#   at line-item level vs order level. It is benign.
# - Differences greater than $0.01 (some up to $300+) represent true data corruption,
#   wrong formula applications, or manual entry typos that skew financial metrics.
# ------------------------------------------------------------------------------
exact_matches = (profit_diff == 0).sum()
penny_rounding = (profit_diff.abs() == 0.01).sum()
severe_profit_mismatches = (profit_diff.abs() > 0.01).sum()

print("=== PROFIT RULE CHECK SUMMARY ===")

print(f"Exact Matches (diff = $0.00):       {exact_matches:,} rows")
print(f"Penny Rounding (diff = +/-$0.01):       {penny_rounding:,} rows")
print(f"Severe Discrepancies (diff > $0.01):        {severe_profit_mismatches:,} rows")


# Filter and display the rows with severe discrepancies (> $0.01) alongside their
# transaction ID, recorded profit, expected profit, and the size of the error.

# WHY:
# Seeing sample rows helps us inspect the patterns and decide whether to recalculate.
# ------------------------------------------------------------------------------
mismatched_profit_df = df_transactions[profit_diff.abs() > 0.01][
    ['Transaction_ID', 'Sales_Amount', 'Cost_Amount', 'Profit']
].copy()

# .abs() takes the absolute value of the difference (so both +0.05 and −0.05 become positive 0.05)
# Creates an independent copy in computer memory.
# This prevents Pandas from throwing a SettingWithCopyWarning when you add new columns to it in the next lines.

mismatched_profit_df['Expected_Profit'] = expected_profit[profit_diff.abs() > 0.01] # New Column
mismatched_profit_df['Difference'] = profit_diff[profit_diff.abs() > 0.01] # New Column

display(mismatched_profit_df.head(5))
```

```python
# Create a two-way frequency table (contingency table / cross-tabulation) between
# Order_Status and Return_Flag.
#
# WHY:
# This gives an instant bird's-eye view of how the two columns correlate.
# For example, does an order marked 'Returned' always have Return_Flag = 'Yes'?
# And does Return_Flag = 'Yes' ever appear under other statuses like 'Completed'?
# ------------------------------------------------------------------------------
print("=== RETURN_FLAG vs ORDER_STATUS CROSS-TABULATION ===")
display(pd.crosstab(df_transactions['Order_Status'], df_transactions['Return_Flag'], margins=True))


# Filter for orders where a return is flagged (Return_Flag == 'Yes'), but the
# Order_Status has not been updated to 'Returned'.
#
# WHY:
# If a customer returned the item, having Order_Status say 'Completed' or 'Shipped'
# causes reporting conflicts:
# - Sales reports counting 'Completed' orders will overstate successful revenue.
# - Fulfillment teams won't see that the item came back into inventory.
# ------------------------------------------------------------------------------
return_mismatches = df_transactions[
    (df_transactions['Return_Flag'] == 'Yes') & (df_transactions['Order_Status'] != 'Returned')
]

print(f"\nTotal Mismatches Found: {len(return_mismatches)} rows")
print("\nBreakdown of Mismatched Order_Status:")
print(return_mismatches['Order_Status'].value_counts())


# Check if these mismatched rows have a valid Return_Reason populated.
#
# WHY:
# If Return_Reason is filled (e.g., 'Defective', 'Wrong Item'), it confirms that
# a real customer return was submitted and processed, proving that 'Return_Flag = Yes'
# is the true ground truth.
# ------------------------------------------------------------------------------
print("\nReturn reasons present in these mismatched rows:")
print(return_mismatches['Return_Reason'].value_counts())
```

```python
# ==============================================================================
# FIX 1: RECALCULATING PROFIT
# ==============================================================================
# Overwrite the 'Profit' column with (Sales_Amount - Cost_Amount).round(2).
#
# WHY:
# - Profit is a deterministic calculated column, not an independent measurement.
# - Leaving severe calculation errors (up to $321 off) and penny roundings will 
#   distort total revenue, profit margin calculations, and financial summaries.
# - Recalculating creates a single source of truth across the whole dataset.
# ==============================================================================
df_transactions['Profit'] = (df_transactions['Sales_Amount'] - df_transactions['Cost_Amount']).round(2)

print("Profit column successfully recalculated across all rows.")


# ==============================================================================
# FIX 2: SYNCHRONIZING ORDER_STATUS WITH RETURN_FLAG
# ==============================================================================
# Where Return_Flag is 'Yes', update Order_Status to 'Returned'.
#
# WHY:
# - All 276 mismatched rows had valid return reasons (Defective, Damaged, etc.),
#   confirming they were indeed returned.
# - In downstream dashboards (e.g., Power BI or executive charts), business analysts
#   often filter by `Order_Status == 'Completed'` to calculate net successful sales.
# - If returned orders remain labeled 'Completed', net sales will be overstated.
# ==============================================================================
df_transactions.loc[df_transactions['Return_Flag'] == 'Yes', 'Order_Status'] = 'Returned'

print("Order_Status successfully synchronized to 'Returned' for all returned orders.")
```
