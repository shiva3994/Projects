##### Count nulls per column and show them in a bar chart. Choose a fix for each column based on data type and % missing (fill, drop or leave blank). 

```python
# Count nulls and percentages
null_count = df_salesdata.isnull().sum()
null_percent = (df_salesdata.isnull().sum()/len(df_salesdata)) * 100

# Plot the Missing Data Bar Chart (Using Column Names on X, Percentages on Y)
plt.figure(figsize=(10, 5))

# null_count.index gives the actual column names
sns.barplot(x=null_count.index, y=null_percent, color="orange")

plt.title("Percentage of Missing Data per Column", fontsize=14)
plt.xlabel("Columns")
plt.ylabel("Percentage Missing (%)")
plt.xticks(rotation=25, ha="right")  # Prevents column names from overlapping
plt.tight_layout()
plt.show()
```
