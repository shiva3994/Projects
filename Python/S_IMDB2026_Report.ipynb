##### Dropping the unwanted columns

```python
cols = ['Url', 'Description', 'Primaryimage','Thumbnails', 'Trailer', 'Endyear','Externallinks']
```

```python
df.drop(cols, axis = 1, inplace = True)
```

```python
df.head()
```

##### Updating date columns format

```python
df.info()
```

```python
#Converting the Startyear column to year format ( int )

# 1. Convert the Releasedate column to actual datetime objects
df['Startyear'] = pd.to_datetime(df['Releasedate'], errors='coerce')

# 2. Extract the year
df['Startyear'] = df['Startyear'].dt.year

# 3. Cast to nullable integer format (Note the CAPITAL 'I')
df['Startyear'] = df['Startyear'].astype('Int64')
```

```python
df['Startyear']
```

```python
#Changing the datatype of Releasedate column to datetime format

df['Releasedate'] = pd.to_datetime(df['Releasedate'])
```

```python
df.info()
```

##### 2. Genres popularity – Which genres appear most often in the Top 250?

```python
df.head()
```

```python
# Split geners in list and explode our data frame to have one genre per row for each movie
df['Genres'] = df['Genres'].str.strip("[]'\"").str.split("', '")

#The explode function is used to transform each element of a list-like to a row, replicating the index values. 
#Here, it is used to create a new row for each genre in the 'Genre' column.
df = df.explode('Genres').reset_index(drop = True)
```

```python
df.info()
```

```python
# Casting column Geners to Category instead of Str

df['Genres'] = df['Genres'].astype('category')
df['Genres'].dtypes
```

```python
df.info()
```

```python
df.head()
```

```python
sns.catplot(y = 'Genres',
            data = df,
            kind = 'count',
            color = 'orange',
            order = df['Genres'].value_counts().index)

plt.title('Count by Genre')
plt.xlabel('Count')
plt.ylabel('Genre')

plt.show()
```

###### Drama, Comedy and Action are top 3 geners in Top 250

##### 3. Budget vs. Rating – Does spending more money make a better movie? 


```python
df.head()
```

```python
df_clean = df[df['Budget'] < 5e8]  
# remove bad currency-mixed values mostly in case of USD

#OR

# More defensible: only drop rows that are BOTH unrealistically high AND non-US
# mask = (df['Budget'] >= 5e8) & (~df['Countriesoforigin'].str.contains("'US'"))
# df_clean = df[~mask]

plt.figure(figsize = (20,8))

sns.scatterplot(x='Budget', y='Averagerating', hue='Genres', data=df_clean, alpha=0.8, s=80)
sns.regplot(x='Budget', y='Averagerating', data=df_clean, scatter = False, color='black', line_kws={'linestyle':'--'})

plt.title('Budget vs. IMDb Rating, colored by Genre')
plt.xlabel('Budget (USD)')
plt.ylabel('Average Rating')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')

plt.show()

# Bucket by budget range for a clearer comparison
# df_clean['Budget_bucket'] = pd.cut(df_clean['Budget'],
#     bins=[0, 10e6, 50e6, 100e6, 200e6, 500e6],
#     labels=['<$10M','$10-50M','$50-100M','$100-200M','$200-500M'])
# df_clean.groupby('Budget_bucket', observed=True)['Averagerating'].mean().plot(kind='bar')
# plt.title('Average Rating by Budget Range')
# plt.ylabel('Average Rating')
# plt.show()
```

##### 7. Country of origin breakdown – Which countries produce the most Top-250-caliber films? 

```python
df.head()
```

```python
# Split countries in list and explode our data frame to have one genre per row for each movie
df['Countriesoforigin'] = df['Countriesoforigin'].str.strip("[]'\"").str.split("', '")

#The explode function is used to transform each element of a list-like to a row, replicating the index values. 
df = df.explode('Countriesoforigin').reset_index(drop = True)

# Casting column Geners to Category instead of Str
df['Genres'] = df['Genres'].astype('category')
df['Genres'].dtypes
```

```python
df.head()
```

```python
plt.figure(figsize=(14, 6))

top_countries = df['Countriesoforigin'].value_counts().head(15)

sns.barplot(x=top_countries.index, y=top_countries.values, color='orange')
plt.title('Top 15 Countries by Number of Top-250 Films')
plt.xlabel('Country')
plt.ylabel('Number of Movies')
plt.show()
```

###### Max movies originated from US, GB and JP
