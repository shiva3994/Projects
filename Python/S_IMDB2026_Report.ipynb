##### 2. Genre popularity – Which genres appear most often in the Top 250?

```python
df.head()
```

```python
# Split geners in list and explode our data frame to have one genre per row for each movie
df['Genres'] = df['Genres'].str.strip("[]'\"").str.split("', '")
```

```python
#The explode function is used to transform each element of a list-like to a row, replicating the index values. 
#Here, it is used to create a new row for each genre in the 'Genre' column.
df = df.explode('Genres').reset_index(drop = True)
```

```python
df.info()
```

```python
# Casting column Geners to Category instead of Str

df['Genres'] = df['Genres'].astype('category')
df['Genres'].dtypes
```

```python
df.info()
```

```python
df.head()
```

```python
sns.catplot(y = 'Genres',
            data = df,
            kind = 'count',
            color = 'orange',
            order = df['Genres'].value_counts().index)

plt.title('Count by Genre')
plt.xlabel('Count')
plt.ylabel('Genre')

plt.show()
```

##### Drama, Comedy and Action are top 3 geners in Top 250
