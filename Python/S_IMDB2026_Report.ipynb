##### Dropping the unwanted columns

```python
cols = ['Url', 'Description', 'Primaryimage','Thumbnails', 'Trailer', 'Endyear','Externallinks']
```

##### Updating date columns format

```python
df.info()
```

```python
#Converting the Startyear column to year format ( int )

df['Startyear'] = pd.to_datetime(df['Startyear'])
df['Startyear'] = df['Startyear'].dt.year
```

```python
#Changing the datatype of Releasedate column to datetime format

df['Releasedate'] = pd.to_datetime(df['Releasedate'])
```

##### 1. Ratings distribution – What does the distribution of Averagerating look like across the Top 250?

```python
ratings_distribution = df['Averagerating'].value_counts()

plt.figure(figsize=(15,6))
sns.barplot(x=ratings_distribution.index, y=ratings_distribution.values, color='orange')
plt.xlabel('Average Rating')
plt.ylabel('Count')
plt.show()

# OR

# plt.figure(figsize=(10,6))
# sns.histplot(df['Averagerating'], bins=30, palette = 'blue')
# plt.title('Distribution of Average Ratings (Top 250 Movies)')
# plt.xlabel('Average Rating')
# plt.ylabel('Count')
# plt.show()
```

##### 2. Genre popularity – Which genres appear most often in the Top 250?

```python
df.head()
```

```python
# Split geners in list and explode our data frame to have one genre per row for each movie
df['Genres'] = df['Genres'].str.strip("[]'\"").str.split("', '")
```

```python
#The explode function is used to transform each element of a list-like to a row, replicating the index values. 
#Here, it is used to create a new row for each genre in the 'Genre' column.
df = df.explode('Genres').reset_index(drop = True)
```

```python
df.info()
```

```python
# Casting column Geners to Category instead of Str

df['Genres'] = df['Genres'].astype('category')
df['Genres'].dtypes
```
